```python
STRING_SESSION = userdata.get('string_ses')
```

```python
STICKER_PACK_NAME = 'DroperPackMafiozkill'
```

```python
async def main():
    async with TelegramClient(StringSession(STRING_SESSION), api_id, api_hash) as client:
        try:
            # get metadata
            pack = await client(functions.messages.GetStickerSetRequest(
                stickerset=types.InputStickerSetShortName(short_name=STICKER_PACK_NAME),
                hash=0
            ))

            sticker_set_id = pack.set.id
            print(f"Stickerpack ID : {sticker_set_id}")

            creator_id = sticker_set_id >> 32
            print(f"Creator ID: {creator_id}")

        except Exception as e:
            print(f"Error: {e}")

await main()
```
